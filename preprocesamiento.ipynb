```python
import os
from nfstream import NFStreamer
import glob
import pandas as pd
```

```python
# Directorios
input_directory = "/home/jproanio/Desktop/Modelo_VAE/Capturas_benignas/"
output_directory = "/home/jproanio/Desktop/Modelo_VAE/Capturas_benignas/"

# Crear el directorio de salida si no existe
os.makedirs(output_directory, exist_ok=True)

# Procesar cada archivo pcap en el directorio de entrada
for filename in os.listdir(input_directory):
    if filename.endswith(".pcap"):  # Verificar si el archivo es un .pcap
        input_path = os.path.join(input_directory, filename)
        output_path = os.path.join(output_directory, f"{os.path.splitext(filename)[0]}.csv")
        
        print(f"Procesando {filename}...")
        
        # Leer el archivo pcap con NFStreamer
        streamer = NFStreamer(source=input_path, statistical_analysis=True)
        df = streamer.to_pandas()
        
        # Agregar columna 'label' con el nombre del archivo sin extensión
        df['label'] = os.path.splitext(filename)[0]
        
        # Guardar el DataFrame como CSV
        df.to_csv(output_path, index=False)
        
        print(f"Archivo CSV creado: {output_path}")

print("Procesamiento completo.")
```
